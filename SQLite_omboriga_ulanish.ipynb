```python
# DB ni o'zgaruvchiga yuklash
db = sql.connect("chinook.db")
cursor = db.cursor()
```

```python

# fetchall metodi yordamida table ni chiqarish
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
cursor.fetchall()
```

```python
cursor.execute("SELECT COUNT (*) FROM employees")
cursor.fetchone() # jadvaldan bitta ma'lumotni chiqarish
```

## Jadvaldan ma'lumotni chiqarish

```python
query = "SELECT CustomerId, FirstName, LastName, Phone, Country FROM customers LIMIT 20"
df = pd.read_sql(query, db, index_col="CustomerId")
df.head()
```

# Tartiblash (ORDER BY) va saralash (DISTINCT)

```python
query = "SELECT CustomerId, FirstName, LastName, Phone, Country FROM customers ORDER BY Country"
df = pd.read_sql(query, db, index_col="CustomerId")
df
```

```python
query = "SELECT DISTINCT Country FROM customers"
cursor.execute(query)
cursor.fetchmany()
```

```python
pd.read_sql(query, db)
```
