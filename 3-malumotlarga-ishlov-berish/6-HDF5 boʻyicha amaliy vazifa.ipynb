3. Saqlab olingan `aqi.csv` faylining oxirgi 5 ta qatorini oʻqing.

Eslatma: fayl nomini `aqi.csv` deb saqlaganimizning sababi AQI – Air Quality Index, yaʼni havo sifati indeksi deganidir. Bizning maʼlumotlarimizda davlatlarning havo sifati haqidagi maʼlumotlar saqlangan!

```python
# 1. Quyidagi link orqali ma'lumotlarni Pandas dataframe'iga yuklab olamiz
import pandas as pd
url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/data_date.csv"
aqi = pd.read_csv(url)
aqi
```

```python
# 1.1 Yuklab olingan ma'lumotlarni aqi.csv fayliga saqlash.
aqi.to_csv('aqi.csv', index=False)
```

```python
# 1.2 Saqlab olingan aqi.csv faylining oxirgi 5 ta qatorini o'qish
aqidf = pd.read_csv('aqi.csv', nrows=5)
```

#2-topshiriq
1. Saqlab olingan `aqi.csv` faylidagi maʼlumotlarning faqat 2024-yilga tegishlisini excel faylga `aqi_2024.xlsx` koʻrinishida saqlab oling.
2. Saqlab olingan `aqi_2024.xlsx` faylidan Oʻzbekistonga tegishli maʼlumotni (qatorlarni) oʻqing.

```python
# 2. aqi.csv faylini chaqirib olamiz
aqidf = pd.read_csv('aqi.csv')
```

```python
# 2.1 'Date' ustunini datetime formatiga o'tkazib olamiz (bu keyinchalik str qiymat bo'lib ma'lumotlarni filtrlashda konflikt bermaslik uchun)
aqidf['Date'] = pd.to_datetime(aqidf['Date'])
```

```python
# 2.2 ma'lumotlarni filtrlash va joriy qilish
aqidf['Date'].dt.year == 2024 # faqat 2024-yildagi ma'lumotlarni qoldirish
aqi2024 = aqidf[aqidf['Date'].dt.year == 2024] # natijani DF ga joriy qilish
aqi2024
```

```python
# 2.3 aqi_2024.xlsx (excel) ko'rinishida saqlab olish
aqi2024.to_excel('aqi_2024.xlsx')
```

```python
# 2.4 aqi_2024.xlsx faylidan O'zbekistonga tegishli ma'lumot (qatorlar)ni filtrlash
uzaqi2024 = pd.read_excel('aqi_2024.xlsx')
uzaqi2024 = uzaqi2024[uzaqi2024['Country'] == 'Uzbekistan']
uzaqi2024.head(10)
```

#3-topshiriq
1. `aqi.csv` faylidagi maʼlumotlarni oʻrganib chiqsangiz, bu faylda 2022–2025-yillar oraligʻidagi havo sifati haqidagi maʼlumotlarni koʻrishingiz mumkin. Hozir sizga vazifa:
  * `aqi_2022.csv`, `aqi_2023.csv`, `aqi_2024.csv`, `aqi_2025.csv`  fayllariga `aqi.csv` faylidagi maʼlumotlarning mos davrlarini saqlang. Misol uchun, `aqi_2022` fayli 2022-yildagi havo sifati haqidagi maʼlumotlarni oʻzida saqlaydi.
  

```python
# 3. AQI 2022 ma'lumotlarini csv faylga yozish
aqidf['Date'].dt.year == 2022
aqi2022 = aqidf[aqidf['Date'].dt.year == 2022]
aqi2022.to_csv('aqi_2022.csv')
```

```python
# 3.1 AQI 2023 ma'lumotlarini csv faylga yozish
aqidf['Date'].dt.year == 2023
aqi2023 = aqidf[aqidf['Date'].dt.year == 2023]
aqi2023.to_csv('aqi_2023.csv')
```

```python
# 3.2 AQI 2024 ma'lumotlarini csv faylga yozish
aqidf['Date'].dt.year == 2024
aqi2024 = aqidf[aqidf['Date'].dt.year == 2024]
aqi2024.to_csv('aqi_2024.csv')
```

```python
# 3.3 AQI 2025 ma'lumotlarini csv faylga yozish
aqidf['Date'].dt.year == 2025
aqi2025 = aqidf[aqidf['Date'].dt.year == 2025]
aqi2025.to_csv('aqi_2025.csv')
```

#4-topshiriq
1. Saqlab olingan `aqi_2022.csv`, `aqi_2023.csv`, `aqi_2024.csv`, `aqi_2025.csv`  fayllarini HDF5 formatida saqlang. Fayl nomini `aqi.h5` deb nomlang. Kalit (key) sifatida fayllarning yillarini bering. Misol uchun, key=2025.

```python
# 4. HDF5 fayl yaratib olish
hdfaqi = pd.HDFStore("aqi.h5")
```

```python
# 4.1 Kalit (key) sifatida fayllarning yillarini berish

hdfaqi.put('aqi2022', aqi2022)
hdfaqi.put('aqi2023', aqi2023)
hdfaqi.put('aqi2024', aqi2024)
hdfaqi.put('aqi2025', aqi2025)

hdfaqi.keys()
```

#5-topshiriq
1. `aqi.h5` faylidan maʼlumotlarni oʻqing va 2025-yilga tegishli maʼlumotlarni yangi dataframeʼga saqlang.
2. `aqi.h5` faylida 2025-yilning qaysi oylari uchun maʼlumotlar bor? Shu oylar roʻyxatini ekranga chiqaring!

```python
# 5. aqi.h5 falidan ma'lumotlarni o'qish va aqi25 df ga saqlash
aqi25 = hdfaqi['aqi2025']
aqi25.sample(10)
```

```python
# 5.1 aqi.h5 faylida 2025-yilning oy (Month) ma'lumotlarini ekranga chiqarish
aqi25['Month'] = aqi25['Date'].dt.month_name()
aqi25['Month'].unique()
```
