```python
import pandas as pd
import sqlite3 as sql
```

```python
db = sql.connect("new.db")
cursor = db.cursor()
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/refs/heads/main/Electricity_Production_By_Source.csv")
df.head()
```

```python
cols = ['country', 'code', 'year', 'coal', 'gas', 'hydro', 'renewable_other', 'solar', 'oil', 'wind', 'nuclear']
df.columns = cols
```

```python
df.to_sql('elect', db)
```

```python
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print(cursor.fetchall())
```

```python
cursor.execute("PRAGMA table_info(elect);")
cursor.fetchall()
```

```python
query = "SELECT * FROM elect WHERE country='Uzbekistan'"
pd.read_sql(query, db)
```
