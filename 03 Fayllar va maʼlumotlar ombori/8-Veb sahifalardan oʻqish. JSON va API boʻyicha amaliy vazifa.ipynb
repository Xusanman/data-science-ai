#1-topshiriq
1. Pandas kutubxonasidan foydalanib, quyida berilgan link ichidan barcha jadvallarni o‘qing va roʻyxat shaklida saqlang.

Link 403 Forbidden (Manzil to'g'ri, lekin kirish taqiqlangan.) xatoligini bermasligi uchun, sahifani avval requests bilan yuklab, keyin pd.read_html() ga HTML matnini bering.

 * https://en.wikipedia.org/wiki/Fossil_fuel_exporters
2. Har bir jadvalning birinchi 1 qatorini ekranga chiqaring.


    

```python
# 1. berilgan link ichidan barcha jadvallarni o'qish
url = "https://en.wikipedia.org/wiki/Fossil_fuel_exporters"
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"} # Wikipedia so'rovni oddiy brauzerdan kelmoqda deb o'ylashi uchun User-Agent beramiz
response = requests.get(url, headers=headers) # web-sahifaga so'rov yuborish
data = pd.read_html(StringIO(response.text)) # HTML tarkibidan jadvallarni o'qib olish
data[1]
```

```python
# 1.2  Har bir jadvalning tartib raqami va 1-qatorini chiqarish
for i, table in enumerate(data):
    print(f"{i+1}-jadval")
    display(table.head(1))
```

#3-topshiriq
1. Quyidagi link orqali json maʼlumotni yuklab oling va uni climate_change_co2.json nomi bilan saqlang.
 * https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/json_data.json
2. Maʼlumotlar boshidan 5 ta qatorni ekranga chiqaring!


```python
# 3. berilgan url manzildan json ma'lumotlarni olish
url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/json_data.json"
co2 = pd.read_json(url) # url dan kelgan json formatdagi ma'lumotlarni o'zgaruvchiga saqlash
co2.to_json('climate_change_co2.json') # o'zgaruvchidagi ma'lumotni json faylga saqlash
co2.head(5)
```

#4-topshiriq
1. Quyida berilgan linkga `get` soʻrovini yuborish orqali `json` maʼlumotlarini `posts.json` fayliga saqlang.
  * https://jsonplaceholder.typicode.com/posts/
2. Ushbu maʼlumotlarni dataframe koʻrinishida saqlang va tasodifiy 5 ta qatorni ekranga chiqaring.

```python
url = "https://jsonplaceholder.typicode.com/posts/"
response = requests.get(url) # url manzildagi json faylga o'tish
df = pd.DataFrame(response.json()) #kelgan json m'alumotlarni DataFrame ga olamiz
df.to_json("posts.json") # ma'lumotlarni json faylga yozamiz
```

```python
df2 = pd.read_json('posts.json') # json falyni o'qish
df2.sample(5) # 5 ta tasodify qatorni chiqarish
```
