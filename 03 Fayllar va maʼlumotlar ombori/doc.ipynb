9. `read_json` - JSON ma'lumotlarni o'qish

10. `read_pickle` - pickle formatidagi fayllarni o'qish

11. `read_sql` - SQL query dan qaytgan ma'lumotlarni o'qish (SQLAlchemy bilan ishlaydi)

17. `index=False` indexni faylga yozmaydi
18. `.to_excel, to_csv, to_html, to_sql` faylga yozish
