#1) Quyida berilgan datasetni yuklab olib, ustun nomlarini koʻring va ularni maʼno jihatdan tushunarli shaklga keltiring. `Unnamed: 0` ustunini tashlab yuboring!

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url)
df.head()
```

```python
# 1.0  Unnamed: 0 column'ni tashlab yuborish
df.drop(['Unnamed: 0'], axis='columns', inplace=True)
```

```python
# 1.1 DF Columns nomlarini o'zgartirish
df.rename(columns={
    'Order ID':'order_id',
    'Customer ID':'customer_id',
    'Category':'category',
    'Item': 'item',
    'Price':'price',
    'Quantity':'quantity',
    'Order Total':'order_total',
    'Order Date':'order_date',
    'Payment Method':'payment_method'
    }, inplace=True)

df.head(10)
```

#2) Maʼlumotlarning umumiy holatini (info(), describe()) tekshiring.

```python
# 2.0 ma'lumotlarning umumiy holatini .info() yordamida tekshirish tekshirish
df.info()
```

```python
# 2.1 ma'lumotlarning umumiy holatini .describe() yordamida tekshirish tekshirish
df.describe()
```

#3) Har bir ustundagi boʻsh qiymatlarni aniqlang va ularni toʻldirish strategiyasini ishlab chiqing.

```python
# 3.0 bo'sh qiymatlarni aniqlash
df.isnull().sum()
```

```python
#3.1 category ustunidagi NaN qiymatlarni tashlab yuborish
df.dropna(subset=['category'], inplace=True)
```

```python
# 3.2 quantity ustunidagi NaN qiymatlarga ishlov berish
df['quantity'] = df['quantity'].fillna(df['order_total'] / df['price'])
```

```python
# 3.3 order_total ustunidagi NaN qiymatlarga ishlov berish
df['order_total'] = df['order_total'].fillna(df['price'] * df['quantity'])
```

```python
# 3.4 payment_method'dagi NaN qiymatlarni to'ldirishda 'Unknown' metodidan foydalandim
df['payment_method'] = df['payment_method'].fillna('Unknown')
```

```python
# 3.5 item 'dagi NaN qiymatlarni to'ldirishda 'Unknown' metodidan foydalandim
df['item'] = df['item'].fillna('Unknown')
```

```python
# 3.7 quantity va order_total ustunlaridagi NaN qiymatlarni tashlab yuborish
df.dropna(subset=['quantity', 'order_total'], inplace=True)
```

```python
df.isna().sum()
```

#4) Har bir ustundagi takroriy qiymatlarni toping va ularga qanday oʻzgartirish kiritish kerakligini hal qiling.

```python
# 4.0 takroriy elementlar mavjud emas
df.duplicated().sum()
```

#5) Maʼlumotlar toʻplamidagi noto‘g‘ri maʼlumotlar (masalan, son turishi kerak bo‘lgan ustunda matn bo‘lsa) mavjudligini tekshiring va to‘g‘rilang.

```python
# 5.0 DF dagi dtype tekshirildi va noodatiy ma'lumot turi mavjud emas
df.dtypes
```

#6) Maʼlumotlar toʻplamidagi **katta/kichik harf** muammolari boʻlgan joylarni toping va bir xil formatga keltiring.

```python
# 6.0 DF ustunlardagi elementlarni katta/kichik harf muammolarni bir xil formatga keltiring
df[['category', 'item', 'payment_method']] = df[['category', 'item', 'payment_method']].apply(lambda s: s.str.strip().str.title())
```

```python
df
```

#7) Maʼlumotlar turini oʻzgartirish.
 * Ustunlarning maʼlumot turlarini tekshiring. Notoʻgʻri turlarga ega boʻlgan ustunlarni toʻgʻri formatga oʻtkazing.
 * `order_date` yoki unga oʻxshash ustunni datetime formatiga oʻtkazing.

```python
# 7.0 order_date ustunini datetime formatiga o'zgartirish
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
```

#8) Mahsulot nomlarida noto‘g‘ri yoki nomaʼlum belgilar (?, NaN, #) mavjudligini tekshiring va ularni tuzating.

```python
# 8.0 mahsulot nomlarida noto'g'ri yoki noma'lum belgilarni olib tashlash
df['item'] = df['item'].str.replace(r'[?#]', '', regex=True).str.strip()
```

#**Maʼlumotlarni filtrlash va qayta ishlash**

#9) Manfiy narx yoki 0 ga teng narx qiymatlarini toping va ularni qanday oʻzgartirish kerakligini hal qiling.

```python
# 9.1 manfiy va 0 ga teng qiymatlarni aniqlash
df[(df['price'] < 0) | (df['price'] == 0)]
```

```python
# 9.2 manfiy sonlarga absolyut qiymat berish orqali 0 dan kichik sonlarni to'g'irlash
df['price'] = np.abs(df['price'])
```

#10) Boʻsh toifalar (kategoriya) mavjudligini tekshiring va toʻgʻri kategoriya berish strategiyasini ishlab chiqing.

```python
# 10.0  category ustununi tahlil qilib item ustunidagi yetishmayotgan elementlarni to'ldirish
mapping = (df.dropna(subset=['category'])
.drop_duplicates('item')
.set_index('item')['category'])

df['category'] = df['category'].fillna(df['item'].map(mapping))
```

#11) Bir xil, lekin turlicha yozilgan qiymatlar (Coffee, cofee, COFEE) mavjudligini tekshiring va tuzating.

```python
# 11.0 bir xil, lekin turlicha yozilgan qiymatlarni tekshirish
df['item'].value_counts()
```

```python
# 11.1 bir xil, lekin turlicha yozilgan qiymatlarni tuzatish
df['item'] = df['item'].str.strip().str.title()
```

#12) `Customer ID` yoki boshqa identifikatorlarning takrorlanishini tekshiring.

```python
# 12.0 order_id yoki boshqa identifikatorlarning takrorlanishini tekshiring
df['customer_id'].duplicated().sum()
```

#13) Mahsulotning ortiqcha taʼriflarini olib tashlang (Water (0.5), Water (1.0), Water → faqat Water qoldirish).

```python
# 13.0 mahsulotlarda ortiqcha tariflar borligini tekshirish
df[df['item'].str.contains(r'\(', na=False)]['item'].unique()
```

```python
# 13.1 mahsulotning ortiqcha taʼriflarini olib tashlash
df['item'] = df['item'].str.replace(r'\s*\(.*?\)', '', regex=True).str.strip()
```

#15) Har bir mijoz oʻrtacha qancha pul sarflaganini hisoblang.

```python
# 15.0 har bir mijoz o'rtacha qancha pul sarflaganini hisoblash
df.groupby('customer_id')['order_total'].mean().sort_values(ascending=False).round(2)
```

#16) Mijozlarning qaysi kunlarda koʻproq xarid qilgani bo‘yicha tahlil qiling.

```python
# 2) Hafta kunlari bo'yicha buyurtmalar soni
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_stats = (df.groupby('weekday')['order_total'].agg(['count']))
weekday_stats
```

#17) Narxi oʻrtacha narxdan ikki barobar yuqori boʻlgan mahsulotlarni toping.

```python
# 17.0 o'rtacha narxlarni hisoblash
mean_price = df['price'].mean()
```

```python
# 17.1 o'rtachadan 2x bo'lgan narxlarni aniqlash
expensive = df[df['price'] > 2 * mean_price]
```

```python
# 17.2 o'rtachadan 2x bo'lgan narxlarni sort qilib chiqarish
expensive = expensive.sort_values('price', ascending=False)
expensive.head(10)
```

#18) Mahsulotlar kategoriyalari bo‘yicha tahlil qiling va eng ko‘p sotiladigan kategoriyalarni aniqlang.

```python
#javobni shu yerda yozing
tahlil = (
    df.groupby('category')
      .agg(orders=('order_id', 'count'))
      .sort_values('orders', ascending=False)
)

tahlil
```

#19) Har bir mijozning oʻrtacha xarid miqdorini (`groupby("customer_id")`) hisoblang.

```python
# 19.0  Har bir mijozning o'rtacha buyurtma summasi
mean = df.groupby('customer_id')['order_total'].mean().round(2)
ortacha_df = mean.reset_index(name='ortacha_xarid')
ortacha_df
```
