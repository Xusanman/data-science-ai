```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/Xusanman/praktikum_datasets/refs/heads/main/world_population.csv', usecols=['country', 'pop2021', 'area'], index_col='country')
df.head()
```

```python
bins = [0, 1000, 10000, 30000, 50000, 100000, 300000, 1500000] # qiymatlarga ajratish
```

```python
population = df.pop2021
```

```python
population
```

```python
guruh = pd.cut(population, bins)
guruh
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Xusanman/praktikum_datasets/refs/heads/main/automobile_data.csv")
df.head()
```

```python
df.price.describe()
```

```python
bins = [0, 1000, 20000, 50000]
```

```python
range = pd.cut(df.price, bins)
range
```

```python
range.value_counts()
```

```python
names = ["arzon", "o'rta", "qimmat"]
range = pd.cut(df.price, bins, labels=names)
range.value_counts()
```

```python
df['narx'] = range
df
```

```python
range = pd.cut(df.price, 5) # qiymatlarni (5 ta) oraliqlarga  ajratish
range.value_counts()
```
